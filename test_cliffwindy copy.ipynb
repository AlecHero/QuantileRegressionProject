```python
import numpy as np
import gymnasium as gym
import importlib
import envs

import learners
import analysis
importlib.reload(analysis)
importlib.reload(learners)
from learners import *
from analysis import *

def get_exp_scheduler(alpha, n_ep_decay, min_val=0.005): return lambda t: max(alpha * 0.5**(t/n_ep_decay), min_val)

params = Params(
    model_name="CliffWindy",
    n_runs=5,
    n_episodes=15_000,
    save_skip=50,
    lr_decay=None,
    seed=123,
    
    N=64,
    alpha=0.2,
    gamma=0.99,
    epsilon=0.0,
    
    nA=None,
    nS=None,
    shape=None,
)
rng = np.random.default_rng(params.seed)
tau = (np.arange(params.N) + 0.5) / params.N
env = gym.make(
    params.model_name,
    shape=(3,5),
)
params = params._replace(nA=int(env.action_space.n))
params = params._replace(nS=int(env.observation_space.n))
params = params._replace(shape=env.unwrapped.shape)
s_init, _ = env.reset()

save_to = "results/results/ALL_"+params.model_name
```

```python
policy, V_pi = PolicyIteration(env, params.gamma)
explorer = EpsilonGreedy(params.epsilon, rng)

returns = MonteCarlo(env, policy, params.gamma, total_episodes=100_000)
tau_returns = np.quantile(np.sort(returns), tau)
```

```python
scheduler = get_exp_scheduler(params.alpha, 1250, min_val=0.00025)

table0 = QR(0, params, policy).train(env, explorer, scheduler)
table1 = QR(1, params, policy).train(env, explorer, scheduler)

qtable = Qlearner(params, policy).train(env, explorer, scheduler)
sarsa = SARSA(params, policy).train(env, explorer, scheduler)
td = TD(params, policy).train(env, explorer, scheduler)
qrtd0 = QRTD(0, params, policy).train(env, explorer, scheduler)
qrtd1 = QRTD(1, params, policy).train(env, explorer, scheduler)
```

```python
# def get_q(table): return table[:,:,s_init].max(2)
def get_q(table): return table[:,:,s_init, policy[s_init].argmax()]
def get_best(table): return get_q(table)[:,-1].mean(0)

plot_objs = [
    (get_best(table0), r"$Z_{\text{QR-0}}$"),
    (get_best(table1), r"$Z_{\text{QR-1}}$"),
    (get_best(qrtd0), r"$Z_{\text{QRTD-0}}$"),
    (get_best(qrtd1), r"$Z_{\text{QRTD-1}}$"),
]
plot_qs = [
    (get_q(table0), "QR-0"),
    (get_q(table1), "QR-1"),
    (get_q(qrtd0), "QRTD-0"),
    (get_q(qrtd1), "QRTD-1"),
    (get_q(qtable), "Q-learning"),
    (get_q(sarsa), "SARSA"),
    (get_q(td), "TD"),
]

LOW_BANDWIDTH = 0.004
bw = LOW_BANDWIDTH*4.0
plot_cdf(plot_objs, tau_returns, tau, linewidths=(3,2), save_to=save_to, returns_name=r"$Z_{\text{True}}$")
plt.xlim(-1,1)
plot_pdf(plot_objs, tau_returns, bandwidth=bw, show_hist=True, linewidth=1.0, N=params.N, hist_alpha=0.2, save_to=save_to)
plt.xlim(-1,1)
plot_qq(plot_objs, tau_returns, save_to=save_to, alpha=0.3, markersize=7, lims=(-1,1))
plot_metrics(plot_qs, tau_returns, save_skip=params.save_skip, save_to=save_to, alphas=[0.2,0.3])
```
