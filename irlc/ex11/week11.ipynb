# Sarsa

$Q(S, A) \leftarrow Q(S,A) + \alpha [R + \gamma Q(S', A') - Q(S,A)]$

```python
# 1.a
A0 = 0
QS0 = [0, 0, 0, 0]
alpha = .9
gamma = 1
R = 0
QS1 = [0, .48, 0, 0]
QS1_possible = [QS0[A0] + alpha * (R + gamma*QS1[Ap] - QS0[A0]) for Ap in range(4)]
QS1_possible
```
