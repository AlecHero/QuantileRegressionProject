```python
import numpy as np
from learners import *
from analysis import *

rng = np.random.default_rng(42)
MoG = MixtureOfGaussians([1/2, 1/2], [-1, 3], [1, 1], rng)


N = 11
plt.style.use("seaborn-v0_8-whitegrid")

tau = (np.arange(N) + 0.5) / N
xs = np.linspace(-5, 7.5, num=200)
qs = MoG.ppf(tau)

tau_edges = np.linspace(0, 1, N+1)
q_edges = MoG.ppf(tau_edges)

plt.plot([q_edges, q_edges], [np.zeros(N+1), MoG.pdf(q_edges)], 'white', linewidth=3.5)
plt.plot(xs, MoG.pdf(xs), color=COLORS[0], linestyle="--", linewidth=2.4)
plt.fill_between(xs, np.zeros(200), MoG.pdf(xs), alpha=0.3, color=COLORS[0])

plt.plot([qs, qs], [0, 0.035], color="black")
plt.plot(qs, np.full_like(qs, 0.035), "o", color="black")

plt.grid(None)
plt.xlim(-4,6)
plt.ylim(0,None)
# plt.savefig("bimodal_tutorial_quantiles.pdf")
plt.axis("off")
plt.tight_layout()
# plt.savefig("bimodal_tutorial_quantiles_toight.pdf", bbox_inches="tight")
plt.savefig("bimodal_tutorial_quantiles.png", bbox_inches="tight", dpi=300)
plt.show()
```
